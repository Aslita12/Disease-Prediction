```python
print("AI DIAGNOSIS ASSISTANT ACTIVATED....")
FEVER=input("do you have fever?'yes'or'no':")
COUGH=input("do you have cough?'yes'or'no':")
HEADACHE=input("do you have headache?'yes'or'no':")
FATIGUE=input("do you feel fatigue/tiredness?'yes'or'no':")
VOMITING=input("do you have vomiting?'yes'or'no':")
CHESTPAIN=input("do you have chestpain?'yes'or'no'")
BREATHING_DIFFICULTY=input("do you have breathing difficulty?'yes'or'no':")
BODY_PAIN=input("do you have BODY PAIN?'yes'or'no':")
EXECISSIVE_THIRST=input("do you feel excessive thirst?'yes'or'no':")
FREQUENT_URINATION=input("do you have frequent urination?'yes'or'no':")
SORE_THROAT=input("do you have sore throat?'yes'or'no':")
COLD_or_SHIVERING=input("do you have cold?'yes'or'no':")
STOMACHPAIN=input("do you have stomach pain?'yes'or'no':")
DIZZINESS=input("do you have dizziness?'yes'or'no':")




print("ANALYSING SYMPTOMS.....")
SEVERITY=""

if COUGH=="yes" and SORE_THROAT=="yes"and FEVER=="yes":
  sneezing=input(f"do you have sneezing?'yes' or 'no'")
  if sneezing=="yes":
    print("COMMAN COLD")
    SEVERITY='MILD'
    print("SEVERITY=MILD")
elif FEVER=="yes" and BODY_PAIN=="yes" and FATIGUE=="yes"and HEADACHE=="yes":
   print("FLU")
   f_s_c=input("do you have fever, shivering,cold?'yes'or 'no'")
   if f_s_c=="yes":
    SEVERITY='MILD'
    print("SEVERITY=MILD")
elif FEVER=="yes" and COUGH=="yes" and FATIGUE=="yes"and BREATHING_DIFFICULTY=="yes":
  print("COVID LIKE INFECTION")
  CP_BD=input("do you have CHEST PAIN, AND BREATHING DIFFICULTY?'yes'or 'no'")
  if CP_BD=="yes":
    SEVERITY='SEVERE'
    print("SEVERITY=SEVERE ")
elif VOMITING=="yes" and STOMACHPAIN=="yes" and FEVER=="yes":
  print("FOOD POISIONING")
  V_D=input("do you have  DIZZINESS AND VOMITING?'yes'or 'no'")
  if V_D=="yes":
    SEVERITY='MODERATE'
    print("SEVERITY=MODERATE")
elif HEADACHE=="yes" and DIZZINESS=="yes" and VOMITING=="yes":
  print("MIGRANE")
  H_S=input("do you have  HEADACHE AND SENSITIVITY TO LIGHT?'yes'or 'no'")
  if H_S=="yes":
    SEVERITY='MODERATE'
    print("SEVERITY=MODERATE")
else:
  print("NOT ABLE TO PREDICT DISEASE AS THE SYMPTOMS DOESNT MATCH!!")


if SEVERITY=='MILD':
  print("rest at home")
elif SEVERITY=='MODERATE':
  print("visit a doctor")
else:
  print("seek emergency care")
```
